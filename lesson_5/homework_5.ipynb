# Задание 1. Извлечение данных из строки (срезы)

Дана строка url = "https://mywebsite.com/api/users/12345/data".

Используя срезы, извлеките из нее подстроки:

"https"

"mywebsite.com"

"12345"

"data"

Подсказка: используйте .find() или .index(), чтобы найти позиции символов /.


```python
url = "https://mywebsite.com/api/users/12345/data"
print('Дана строка', url)
# Разбиваем URL на части
parts = url.split('/')
# parts = ['https:', '', 'mywebsite.com', 'api', 'users', '12345', 'data']

protocol = parts[0].rstrip(':')          # "https"
domain = parts[2]                        # "mywebsite.com"
user_id = parts[5]                       # "12345"
data_type = parts[6]                     # "data"

print('Протокол:', protocol)
print('Домен:', domain)
print('ID пользователя:', user_id)
print('Тип данных:', data_type)
```


# Задание 2. Распаковка значений

Дан список user_info = ["Alice", 30, "Engineer"].

Используя распаковку, присвойте значения его элементов трем переменным name, age, profession.

Выведите на экран сообщение: "Имя: Alice, Возраст: 30, Профессия: Engineer".


```python
user_info = ["Alice", 30, "Engineer"]
print('Дан список:', user_info)

name, age, profession = user_info

# Исправляем опечатку в слове "Распаковали"
print('Распаковали список:', name, age, profession)
print('Имя:', name, 'Возраст:', age, 'Профессия:', profession)
```


# Задание 3. Форматирование строк (f-strings)

Запросите у пользователя его имя и возраст.

Используя f-строку, выведите сообщение: "Привет, {имя}! Тебе {возраст} лет. Через 5 лет тебе будет {возраст+5} лет.".

Попробуйте сделать то же самое с методами .format() и %s.

```python
# Запрашиваем имя
user_name = input('Введите ваше имя: ')
print(f'Вы ввели имя: {user_name}')

# Запрашиваем возраст (преобразуем в число)
user_age = int(input('Введите ваш возраст: '))
print(f'Вы ввели возраст: {user_age}')

# Выводим сообщение с f-строкой
print(f'Привет, {user_name}! Тебе {user_age} лет. Через 5 лет тебе будет {user_age + 5} лет.')
```


# Задание 4. Очистка пользовательского ввода

Напишите программу, которая запрашивает у пользователя email.

Удалите из введенной строки все начальные и конечные пробелы (используйте .strip()).

Убедитесь, что строка написана в нижнем регистре (используйте .lower()).

Проверьте, содержит ли строка символ @. Если да, выведите "Email корректен", если нет — "Email не содержит @".

```python
# 1. Запрашиваем email
email = input('Введите ваш email: ')
print(f'Вы ввели: "{email}"')

# 2. Удаляем пробелы по краям
email_cleaned = email.strip()
print(f'После удаления пробелов: "{email_cleaned}"')

# 3. Приводим к нижнему регистру
email_lower = email_cleaned.lower()
print(f'В нижнем регистре: "{email_lower}"')

# 4. Проверяем наличие @
if '@' in email_lower:
    print('Email корректен')
else:
    print('Email не содержит @')
```
