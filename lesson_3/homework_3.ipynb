# Задание 1. Калькулятор площади прямоугольника
Условие:

Создайте переменные width и height и присвойте им числовые значения (например, 5 и 10).

Вычислите площадь прямоугольника по формуле area = width * height.

Выведите на экран сообщение: "Площадь прямоугольника со сторонами 5 и 10 равна 50".

Усложнение: Запросите значения width и height у пользователя с помощью input(), преобразуйте их в числа (int или float) и выведите результат.



```python

# Базовый вариант
width = 5
height = 10
area = width * height

print(f'Площадь прямоугольника со сторонами {width} и {height} равна {area}')

# Усложнение: с вводом от пользователя
width = float(input('Введите ширину прямоугольника: '))
height = float(input('Введите высоту прямоугольника: '))
area = width * height

print(f'Площадь прямоугольника со сторонами {width} и {height} равна {area}')
```

# Задание 2. Проверка на четность
Условие:

Запросите у пользователя целое число.

Используя оператор % (остаток от деления), проверьте, является ли число четным.

Выведите на экран True, если число четное, и False, если нечетное.

Подсказка: число четное, если number % 2 == 0.


```python
number = int(input('Введите целое число: '))

# Проверяем четность
is_even = number % 2 == 0

print(is_even)

# Вариант с пояснением
if number % 2 == 0:
    print(f'Число {number} четное - True')
else:
    print(f'Число {number} нечетное - False')
```

# Задание 3. Конкатенация и приведение типов
Условие:

Создайте переменные: product_name = "Ноутбук", price = 75000, quantity = 3.

Создайте переменную total_price и вычислите общую стоимость (price * quantity). Обратите внимание, что price и quantity — числа.

Сформируйте строку вида: "Вы купили Ноутбук в количестве 3 шт. на сумму 225000 руб.".

Для этого вам нужно будет преобразовать число total_price в строку с помощью str().

```python
product_name = "Ноутбук"
price = 75000
quantity = 3
total_price = price * quantity

# Способ 1: f-строка (рекомендуемый)
print(f'Вы купили {product_name} в количестве {quantity} шт. на сумму {total_price} руб.')

# Способ 2: с помощью str()
print('Вы купили ' + product_name + ' в количестве ' + str(quantity) + ' шт. на сумму ' + str(total_price) + ' руб.')

# Способ 3: .format()
print('Вы купили {} в количестве {} шт. на сумму {} руб.'.format(product_name, quantity, total_price))
```

# Задание 4. Сравнение строк
Условие:

Запросите у пользователя его имя.

Сравните его с каким-либо именем (например, "admin") и выведите результат сравнения (True или False).

Важно: Сравнение должно быть регистронезависимым. Используйте метод .lower() для приведения строк к одному регистру.



```python
user_name = input('Введите ваше имя: ')

# Приводим к нижнему регистру и сравниваем
is_admin = user_name.lower() == 'admin'

print(is_admin)

# Вариант с пояснением
if user_name.lower() == 'admin':
    print('True - вы администратор')
else:
    print('False - вы обычный пользователь')
```
