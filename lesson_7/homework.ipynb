# Условие задачи
Вы работаете над системой автоматизации для интернет-магазина. Вам нужно написать программу, которая:

Хранит список товаров в виде словаря, где ключ — название товара, значение — его цена.

Позволяет пользователю добавлять товары в корзину.

Выводит итоговую стоимость корзины.

Программа должна работать через цикл while и использовать функции для выполнения основных операций.

# Критерии проверки

✅ Код читаемый, переменные названы понятно

```python
products = {
    "яблоко": 50,
    "банан": 30,
    "молоко": 80,
    "хлеб": 40,
    "сыр": 150
}

cart = {}

def show_products():
    print("\nДоступные товары:")
    for product, price in products.items():
        print(f"{product} - {price} руб.")
    print()

def add_to_cart(product_name, quantity):
    if product_name not in products:
        print(f"Товар '{product_name}' не найден")
        return
    cart[product_name] = cart.get(product_name, 0) + quantity
    print(f"Товар '{product_name}' добавлен в корзину ({quantity} шт.)")

def show_cart():
    if not cart:
        print("\nКорзина пуста.")
        return
    print("\nВаша корзина:")
    total = 0
    for product, quantity in cart.items():
        price = products[product]
        cost = price * quantity
        total += cost
        print(f"{product}: {quantity} шт. - {cost} руб.")
    print(f"Итого: {total} руб.\n")

def clear_cart():
    cart.clear()
    print("Корзина очищена.")

# Основной цикл программы
print("Доступные команды:")
print("list - показать все товары")
print("add - добавить товар в корзину")
print("show - показать корзину")
print("clear - очистить корзину")
print("exit - выход\n")

while True:
    command = input("Введите команду: ").strip().lower()
    
    if command == "exit":
        print("До свидания!")
        break
    
    elif command == "list":
        show_products()
    
    elif command == "add":
        product = input("Введите название товара: ").strip().lower()
        try:
            quantity = int(input("Введите количество: "))
            add_to_cart(product, quantity)
        except ValueError:
            print("Ошибка: нужно ввести число!")
    
    elif command == "show":
        show_cart()
    
    elif command == "clear":
        clear_cart()
    
    else:
        print("Неизвестная команда. Попробуйте снова.")
```
