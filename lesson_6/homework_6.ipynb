# Задание 1. Игра "Угадай число" (if-elif-else)

Задайте в переменной secret_number любое число от 1 до 10.

Запросите у пользователя число.

Используя условия, выведите:

"Поздравляю! Вы угадали!", если число совпадает.

"Загаданное число больше.", если число пользователя меньше.

"Загаданное число меньше.", если число пользователя больше.

Дополнительное условие: Проверьте, что пользователь ввел именно число, а не строку (используйте .isdigit()).

```python
import random

# Генерируем случайное число от 0 до 100
secret_number = random.randint(0, 100)
# print(secret_number)  # Закомментировано для игры

# Запрашиваем у пользователя число
user_number = int(input('Введите число: '))

# Сравниваем
if user_number == secret_number:
    print('Поздравляю! Вы угадали!')
elif user_number < secret_number:
    print('Загаданное число больше.')
else:  # user_number > secret_number
    print('Загаданное число меньше.')
```
