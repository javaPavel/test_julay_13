# Задание 1. Работа со списком студентов
Условие:

Создайте список students = ["Иван", "Петр", "Мария", "Анна"].

Добавьте в конец списка нового студента "Ольга" с помощью .append().

Вставьте студента "Сергей" на позицию с индексом 1 с помощью .insert().

Узнайте индекс студента "Мария" с помощью .index().

Выведите на экран последнего студента в списке, используя отрицательный индекс.

Удалите студента "Петр" с помощью .pop() или .remove().


```python
# 1. Создаем список
students = ["Иван", "Петр", "Мария", "Анна"]
print('Начальный список:', students)

# 2. Добавляем в конец
students.append('Ольга')
print('После добавления Ольги:', students)

# 3. Вставляем на позицию 1
students.insert(1, 'Сергей')
print('После вставки Сергея:', students)

# 4. Узнаем индекс Марии
maria_index = students.index('Мария')
print('Индекс Марии:', maria_index)

# 5. Выводим последнего студента
print('Последний студент:', students[-1])

# 6. Удаляем Петра (два способа)
# Способ 1: через remove
students.remove('Петр')
print('После удаления Петра (remove):', students)

# Способ 2: через pop с индексом (закомментирован)
# students.pop(1)  # Петр теперь на позиции 1
# print('После удаления Петра (pop):', students)
```

# Задание 2. Уникальные элементы (кортеж и множество)
Условие:

Дан список с повторяющимися числами: numbers = [1, 2, 2, 3, 4, 4, 4, 5].

Создайте множество (set) из этого списка, чтобы оставить только уникальные числа.

Преобразуйте полученное множество обратно в список и выведите его. Какой порядок элементов получился?

Создайте кортеж из трех любых элементов. Попытайтесь изменить один из элементов кортежа (например, my_tuple[0] = 10), чтобы увидеть ошибку.


```python
# Часть 1: Работа с множеством
numbers = [1, 2, 2, 3, 4, 4, 4, 5]
print('Исходный список:', numbers)

# Создаем множество (удаляем дубликаты)
unique_set = set(numbers)
print('Множество (уникальные значения):', unique_set)

# Преобразуем обратно в список
unique_list = list(unique_set)
print('Список уникальных значений:', unique_list)
print('Порядок элементов может отличаться от исходного!')

# Часть 2: Работа с кортежем
my_tuple = (10, 20, 30)
print('Кортеж:', my_tuple)

# Попытка изменить кортеж (вызовет ошибку)
try:
    my_tuple[0] = 10
    print('Кортеж изменен!')  # Эта строка не выполнится
except TypeError as e:
    print(f'Ошибка: {e}')
    print('Кортежи неизменяемы!')

# Создаем кортеж из одного элемента
single_tuple = (42,)
print(f'Кортеж из одного элемента: {single_tuple}')
```


# Задание 3. Словарь "Библиотека"
Условие:

Создайте словарь book = {"title": "Война и мир", "author": "Толстой", "year": 1869}.

Добавьте в словарь новый ключ "genre" со значением "Роман".

Измените значение ключа "year" на 1867 (перепроверьте дату).

Выведите на экран все ключи словаря с помощью .keys().

Проверьте, есть ли в словаре ключ "publisher" с помощью оператора in.


```python

# 1. Создаем словарь
book = {"title": "Война и мир", "author": "Толстой", "year": 1869}
print('Создали словарь:', book)

# 2. Добавляем новый ключ
book['genre'] = 'Роман'
print('Добавили жанр:', book)

# 3. Изменяем значение года
book['year'] = 1867
print('Изменили год:', book)

# 4. Выводим все ключи
print('Все ключи словаря:', book.keys())

# 5. Проверяем наличие ключа "publisher"
has_publisher = 'publisher' in book
print('Есть ли ключ "publisher"?', has_publisher)

# Вариант с условием
if 'publisher' in book:
    print('Ключ "publisher" найден!')
else:
    print('Ключа "publisher" нет в словаре.')
```


# Задание 4. Анализ строки
Условие:

Дана строка: "apple,banana,apple,orange,banana,grape".

Преобразуйте эту строку в список, используя метод .split(',').

Преобразуйте полученный список в множество, чтобы убрать дубликаты.

Снова преобразуйте множество в список.

Отсортируйте этот новый список с помощью функции sorted() (она возвращает новый отсортированный список).

Склейте элементы отсортированного списка обратно в строку через запятую с помощью ",".join(). Выведите результат.


```python

# 1. Исходная строка
text = "apple,banana,apple,orange,banana,grape"
print('Исходная строка:', text)

# 2. Преобразуем в список
fruits_list = text.split(',')
print('Список:', fruits_list)

# 3. Преобразуем в множество (убираем дубликаты)
fruits_set = set(fruits_list)
print('Множество (уникальные):', fruits_set)

# 4. Снова в список
unique_list = list(fruits_set)
print('Список уникальных:', unique_list)

# 5. Сортируем с помощью sorted()
sorted_list = sorted(unique_list)
print('Отсортированный список:', sorted_list)
print('Исходный список не изменился:', unique_list)

# 6. Склеиваем обратно в строку
result = ",".join(sorted_list)
print('Итоговая строка:', result)

# Альтернативный вариант с пробелом после запятой
result_with_space = ", ".join(sorted_list)
print('С пробелом после запятой:', result_with_space)
```
